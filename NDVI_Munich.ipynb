```python
# Sentinel-2 NDVI Analysis — Munich, Germany

**Author:** Dharuna S  
**Date:** March 2026  
**Tools:** Python, Google Earth Engine, geemap

## What this notebook does
This analysis uses Sentinel-2 imagery to calculate 
and visualise the NDVI index over Munich, Germany. I tried out cloud filtering, band mathematics, 
and multi-layer visualisation using the Google Earth Engine Python API.

## Data
- **Sensor:** Sentinel-2 MSI (Level-2A surface reflectance)
- **Dataset:** COPERNICUS/S2_SR_HARMONIZED
- **Period:** June — September 2024
- **Area:** Munich city centre, 20km radius
```

```python
#Loading a map of Munich, with a cloud cover filter of 10 percent
```

```python
import ee
import geemap

ee.Authenticate()
ee.Initialize(project='my-project-1535639102467')

region = ee.Geometry.Point([11.576, 48.137]).buffer(20000)
image = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
           .filterBounds(region)
           .filterDate('2024-06-01', '2024-09-01')
           .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 10))
           .first())

Map = geemap.Map()
Map.centerObject(region, 10)
Map.addLayer(image, {'bands': ['B4','B3','B2'], 'min':0, 'max':3000}, 'Munich RGB')
Map
```

```python
# Calculating the NDVI index
```

```python
ndvi = image.normalizedDifference(['B8', 'B4'])
```

```python
# Calculate NDVI
ndvi = image.normalizedDifference(['B8', 'B4'])

# Add it to the map
Map.addLayer(ndvi, {
    'min': -0.2,
    'max': 0.8,
    'palette': ['brown', 'yellow', 'green']
}, 'NDVI')

Map
```
